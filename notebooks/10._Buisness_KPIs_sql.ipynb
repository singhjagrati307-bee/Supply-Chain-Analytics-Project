# 3. Business KPIs

## Objective

Calculate key Supply Chain KPIs to evaluate order fulfillment performance and delivery efficiency.

```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect("supply_chain.db")

print("✅ Connected to SQLite Database Successfully!")
```

```python
## KPI 1 - Total Orders
pd.read_sql("""
-- Business Question:
-- How many unique customer orders were placed?

SELECT
    COUNT(DISTINCT order_id) AS total_orders
FROM fact_aggregate;
""", conn)
```

```python
## KPI 2 – Total Order Lines
pd.read_sql("""
-- Business Question:
-- How many order lines were processed?

SELECT
    COUNT(*) AS total_order_lines
FROM fact_order_line;
""", conn)
```

```python
## KPI 3 – On-Time Delivery %
pd.read_sql("""
-- Business Question:
-- What percentage of order lines were delivered on time?

SELECT
    ROUND(AVG("On Time") * 100, 2) AS on_time_delivery_percentage
FROM fact_order_line;
""", conn)
```

```python
## KPI 4 – In-Full Delivery %
pd.read_sql("""
-- Business Question:
-- What percentage of order lines were delivered in full?

SELECT
    ROUND(AVG("In Full") * 100, 2) AS in_full_delivery_percentage
FROM fact_order_line;
""", conn)
```

```python
## KPI 5 – OTIF %
pd.read_sql("""
-- Business Question:
-- What percentage of order lines were delivered On Time In Full?

SELECT
    ROUND(AVG("On Time In Full") * 100, 2) AS otif_percentage
FROM fact_order_line;
""", conn)
```

```python
## KPI 6 – Line Fill Rate = (Number of order lines delivered in full ÷ Total order lines) × 100
pd.read_sql("""
SELECT
    ROUND(
        SUM("In Full") * 100.0 / COUNT(*),
        2
    ) AS line_fill_rate
FROM fact_order_line;
""", conn)
```

```python
## KPI 7 – Volume Fill Rate
## (Total Delivered Quantity ÷ Total Ordered Quantity) × 100
pd.read_sql("""
SELECT
    ROUND(
        SUM(delivery_qty) * 100.0 /
        SUM(order_qty),
        2
    ) AS volume_fill_rate
FROM fact_order_line;
""", conn)
```
