```python
import pandas as pd

customers = pd.read_csv("../data/dim_customers.csv")
products = pd.read_csv("../data/dim_products.csv")
targets = pd.read_csv("../data/dim_targets_orders.csv")

order_line = pd.read_csv("../data/fact_order_line.csv")
aggregate = pd.read_csv("../data/fact_aggregate.csv")
```

```python
print("Customers Shape :", customers.shape)
print("Products Shape :", products.shape)
print("Order Line Shape :", order_line.shape)
print("Aggregate Shape :", aggregate.shape)
print("Targets Shape :", targets.shape)
```

```python
print("=" * 60)
print("Customers Data Types")
print("=" * 60)
print(customers.dtypes)

print("\n" + "=" * 60)
print("Products Data Types")
print("=" * 60)
print(products.dtypes)

print("\n" + "=" * 60)
print("Order Line Data Types")
print("=" * 60)
print(order_line.dtypes)

print("\n" + "=" * 60)
print("Aggregate Data Types")
print("=" * 60)
print(aggregate.dtypes)

print("\n" + "=" * 60)
print("Targets Data Types")
print("=" * 60)
print(targets.dtypes)
```

```python
# Convert date columns in order_line table

order_line["order_placement_date"] = pd.to_datetime(
    order_line["order_placement_date"],
    dayfirst=True
)

order_line["agreed_delivery_date"] = pd.to_datetime(
    order_line["agreed_delivery_date"],
    dayfirst=True
)

order_line["actual_delivery_date"] = pd.to_datetime(
    order_line["actual_delivery_date"],
    dayfirst=True
)

# Convert date column in aggregate table

aggregate["order_placement_date"] = pd.to_datetime(
    aggregate["order_placement_date"],
    dayfirst=True
)

print("Date columns converted successfully.")
```

```python
# verify Date conversion
print("=" * 60)
print("Order Line Data Types After Conversion")
print("=" * 60)
print(order_line.dtypes)

print("\n" + "=" * 60)
print("Aggregate Data Types After Conversion")
print("=" * 60)
print(aggregate.dtypes)
```

```python
# check duplicate records in all tables 
print("=" * 60)
print("Duplicate Records")
print("=" * 60)

print("Customers :", customers.duplicated().sum())

print("Products :", products.duplicated().sum())

print("Order Line :", order_line.duplicated().sum())

print("Aggregate :", aggregate.duplicated().sum())

print("Targets :", targets.duplicated().sum())
```

```python
## Remove Duplicate Records
customers.drop_duplicates(inplace=True)

products.drop_duplicates(inplace=True)

order_line.drop_duplicates(inplace=True)

aggregate.drop_duplicates(inplace=True)

targets.drop_duplicates(inplace=True)

print("Duplicate records removed successfully.")
```

```python
## Verify Duplicate Removal
print("=" * 60)
print("Duplicate Records After Cleaning")
print("=" * 60)

print("Customers :", customers.duplicated().sum())

print("Products :", products.duplicated().sum())

print("Order Line :", order_line.duplicated().sum())

print("Aggregate :", aggregate.duplicated().sum())

print("Targets :", targets.duplicated().sum())
```

```python
## Check Missing Values
print("=" * 60)
print("Missing Values in Customers")
print("=" * 60)
print(customers.isnull().sum())

print("\n" + "=" * 60)
print("Missing Values in Products")
print("=" * 60)
print(products.isnull().sum())

print("\n" + "=" * 60)
print("Missing Values in Order Line")
print("=" * 60)
print(order_line.isnull().sum())

print("\n" + "=" * 60)
print("Missing Values in Aggregate")
print("=" * 60)
print(aggregate.isnull().sum())

print("\n" + "=" * 60)
print("Missing Values in Targets")
print("=" * 60)
print(targets.isnull().sum())
```

```python
## Validate Missing Values
if (
    customers.isnull().sum().sum() == 0
    and products.isnull().sum().sum() == 0
    and order_line.isnull().sum().sum() == 0
    and aggregate.isnull().sum().sum() == 0
    and targets.isnull().sum().sum() == 0
):
    print("No missing values found in any dataset.")
else:
    print("Missing values detected. Data cleaning required.")
```

```python
## Verify Final Dataset Dimensions
print("=" * 60)
print("Final Dataset Shapes")
print("=" * 60)

print("Customers :", customers.shape)
print("Products :", products.shape)
print("Order Line :", order_line.shape)
print("Aggregate :", aggregate.shape)
print("Targets :", targets.shape)
```
