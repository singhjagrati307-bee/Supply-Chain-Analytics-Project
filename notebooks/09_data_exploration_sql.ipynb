# 2. Data Exploration

## Objective

The purpose of this section is to explore the supply chain database and understand the size, structure, and distribution of the data before performing KPI analysis.

The analysis covers:

- Customer distribution
- Product distribution
- Order statistics
- Geographic distribution
- Currency distribution

```python
import pandas as pd
```

```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect("supply_chain.db")

print("✅ Connected to SQLite Database Successfully!")
```

```python
## view all tables in the database 
pd.read_sql("""
SELECT name
FROM sqlite_master
WHERE type='table';
""", conn)
```

```python
## Total Customers in the database 
pd.read_sql("""
SELECT
    COUNT(*) AS total_customers
FROM dim_customers;
""", conn)
```

```python
## Customer Distribution by City
pd.read_sql("""
SELECT
    city,
    COUNT(*) AS total_customers
FROM dim_customers
GROUP BY city
ORDER BY total_customers DESC;
""", conn)
```

```python
## Customer Distribution by Currency
pd.read_sql("""
SELECT
    currency,
    COUNT(*) AS total_customers
FROM dim_customers
GROUP BY currency;
""", conn)
```

```python
## Total Products
pd.read_sql("""
SELECT
    COUNT(*) AS total_products
FROM dim_products;
""", conn)
```

```python
## Products by Category
pd.read_sql("""
SELECT
    category,
    COUNT(*) AS total_products
FROM dim_products
GROUP BY category
ORDER BY total_products DESC;
""", conn)

```

```python
pd.read_sql("""
SELECT
    product_id,
    product_name,
    category
FROM dim_products
ORDER BY category, product_name;
""", conn)
```

```python
## Total Orders
pd.read_sql("""
SELECT
    COUNT(DISTINCT order_id) AS total_orders
FROM fact_aggregate;
""", conn)
```

```python
## Total Order Lines
pd.read_sql("""
SELECT
    COUNT(*) AS total_order_lines
FROM fact_order_line;
""", conn)
```

```python
## Preview Aggregate Table
pd.read_sql("""
SELECT *
FROM fact_aggregate
LIMIT 10;
""", conn)
```

```python
## Preview Order Line Table
pd.read_sql("""
SELECT *
FROM fact_order_line
LIMIT 10;
""", conn)
```

```python
## Check for Missing Customer Names
pd.read_sql("""
SELECT *
FROM dim_customers
WHERE customer_name IS NULL;
""", conn)
```

```python
## Check for missing Product Names 
pd.read_sql("""
SELECT *
FROM dim_products
WHERE product_name IS NULL;
""", conn)
```

```python
## Count Orders by Customer
pd.read_sql("""
SELECT
    customer_id,
    COUNT(DISTINCT order_id) AS total_orders
FROM fact_aggregate
GROUP BY customer_id
ORDER BY total_orders DESC;
""", conn)
```

```python
## Count Products Ordered
pd.read_sql("""
SELECT
    product_id,
    COUNT(*) AS total_order_lines
FROM fact_order_line
GROUP BY product_id
ORDER BY total_order_lines DESC;
""", conn)
```
