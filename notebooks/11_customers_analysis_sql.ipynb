# 4. Customer Analysis

## Objective

Analyze customer purchasing behavior and service performance.

This section identifies:

- Top customers by order value
- Customer-wise revenue
- OTIF performance
- On-Time Delivery %
- In-Full Delivery %

```python
import pandas as pd 
import sqlite3
# Connect to the SQLite database
conn = sqlite3.connect("../data/supply_chain.db")

print("✅ Connected to SQLite Database Successfully!")
```

```python
## Bottom 5 Customers by Revenue
pd.read_sql("""
SELECT

c.customer_name,

c.city,

SUM(
f.delivery_qty*p.price_INR
) AS revenue

FROM fact_order_line f

JOIN dim_customers c
ON f.customer_id=c.customer_id

JOIN dim_products p
ON f.product_id=p.product_id

GROUP BY
c.customer_name,
c.city

ORDER BY revenue ASC

LIMIT 5;

""", conn)
```

```python
## Top five customers by order value
pd.read_sql("""
SELECT
    c.customer_id,
    c.customer_name,
    c.city,

    ROUND(
        SUM(f.delivery_qty * p.price_INR),
        2
    ) AS order_value

FROM fact_order_line AS f

JOIN dim_customers AS c
ON f.customer_id = c.customer_id

JOIN dim_products AS p
ON f.product_id = p.product_id

GROUP BY
    c.customer_id,
    c.customer_name,
    c.city

ORDER BY order_value DESC

LIMIT 5;
""", conn)
```

```python
import pandas as pd
pd.read_sql("""
SELECT COUNT(*)
FROM dim_customers;
""",conn)           
```

```python
pd.read_sql("""
SELECT
    c.customer_id,
    c.customer_name,
    c.city,
    SUM(f.delivery_qty * p.price_INR) AS total_revenue
FROM fact_order_line f
JOIN dim_customers c
    ON f.customer_id = c.customer_id
JOIN dim_products p
    ON f.product_id = p.product_id
GROUP BY
    c.customer_id,
    c.customer_name,
    c.city
ORDER BY total_revenue DESC;
""",conn)
```

```python
pd.read_sql("""
SELECT
    customer_name,
    COUNT(*) AS cnt
FROM dim_customers
GROUP BY customer_name
HAVING COUNT(*) > 1;
""",conn)
```

```python
## Top 5 Customers with OTIF %
pd.read_sql("""
SELECT

    c.customer_id,

    c.customer_name,

    c.city,

    ROUND(
        SUM(f.delivery_qty * p.price_INR),
        2
    ) AS order_value,

    ROUND(
        AVG(f."On Time In Full") * 100,
        2
    ) AS otif_percentage

FROM fact_order_line f

JOIN dim_customers c
ON f.customer_id = c.customer_id

JOIN dim_products p
ON f.product_id = p.product_id

GROUP BY

    c.customer_id,

    c.customer_name,

    c.city

ORDER BY order_value DESC

LIMIT 5;
""", conn)
```

```python
## Top 5 Customers with OTIF, OT and IF
pd.read_sql("""
SELECT

    c.customer_id,

    c.customer_name,

    c.city,

    ROUND(
        SUM(f.delivery_qty * p.price_INR),
        2
    ) AS order_value,

    ROUND(
        AVG(f."On Time In Full") * 100,
        2
    ) AS otif_percentage,

    ROUND(
        AVG(f."On Time") * 100,
        2
    ) AS on_time_percentage,

    ROUND(
        AVG(f."In Full") * 100,
        2
    ) AS in_full_percentage

FROM fact_order_line f

JOIN dim_customers c
ON f.customer_id = c.customer_id

JOIN dim_products p
ON f.product_id = p.product_id

GROUP BY

    c.customer_id,

    c.customer_name,

    c.city

ORDER BY order_value DESC

LIMIT 5;
""", conn)
```
