```python
import sqlite3
import pandas as pd

# Connect to SQLite
conn = sqlite3.connect("supply_chain.db")

# Read CSV files
customers = pd.read_csv("../data/dim_customers.csv")
products = pd.read_csv("../data/dim_products.csv")
targets = pd.read_csv("../data/dim_targets_orders.csv")

order_line = pd.read_csv("../data/fact_order_line.csv")
aggregate = pd.read_csv("../data/fact_aggregate.csv")

# Import into SQLite
customers.to_sql("dim_customers", conn, if_exists="replace", index=False)
products.to_sql("dim_products", conn, if_exists="replace", index=False)
targets.to_sql("dim_targets_orders", conn, if_exists="replace", index=False)
aggregate.to_sql("fact_aggregate", conn, if_exists="replace", index=False)
order_line.to_sql("fact_order_line", conn, if_exists="replace", index=False)

print("✅ All CSV files imported successfully!")

# Close the connection LAST
conn.close()
```
